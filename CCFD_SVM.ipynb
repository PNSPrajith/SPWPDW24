```python
#importing all the libraries

import pandas as pd
from sklearn.metrics import classification_report, auc, precision_recall_curve, confusion_matrix
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.preprocessing import StandardScaler
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load data
df = pd.read_csv(r"C:\Users\Naga Sai\Desktop\Python\creditcard.csv") #creating a dataframe
x = df.drop('Class', axis=1) # another way of creating dataframe with all column other than 'class'
y = df['Class']
```

```python
# Scale features
scaler = StandardScaler()
x_scaled = scaler.fit_transform(x)
```

```python
# Train/test split
x_train, x_test, y_train, y_test = train_test_split(x_scaled, y, test_size=0.2, stratify=y, random_state=32)
```

```python
# SVM with class weight
svm = SVC(kernel='linear', class_weight='balanced', probability=True)
svm.fit(x_train, y_train)
```

```python
# Predict and evaluate
y_pred = svm.predict(x_test)
scores = svm.decision_function(x_test)
```

```python
print(classification_report(y_test, y_pred))
```

```python
# AUC-PR
precision, recall, thresholds = precision_recall_curve(y_test, scores)
print('AUC (PR curve):', auc(recall, precision))
```

```python
# Precision-Recall curve
plt.plot(recall, precision, marker='.')
plt.xlabel('Recall')
plt.ylabel('Precision')
plt.title('Precision-Recall Curve')
plt.grid()
plt.show()
```

```python
# Confusion matrix
cm = confusion_matrix(y_test, y_pred)
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues')
plt.title("Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.show()
```

```python
# Class counts
print("Predicted class distribution:\n", pd.Series(y_pred).value_counts())
print("True class distribution:\n", pd.Series(y_test).value_counts())
```
